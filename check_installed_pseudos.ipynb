```python
import aiida
from aiida.orm import Group, QueryBuilder
from aiida.plugins import DataFactory

try:
    aiida.load_profile()
    print("AiiDA Profile loaded successfully.")
except Exception as e:
    print(f"Profile already loaded or: {e}")


try:
   
    group = Group.collection.get(label="psf_family")
    print(f"Family 'psf_family' found (PK: {group.pk})")
    print("-" * 40)

    for node in group.nodes:
        print(f"* Element: {node.element}")
        print(f"   PK: {node.pk}")
        print(f"   Filename: {node.filename}")
        
     
        content = node.get_content().splitlines()
        print(f"   Header Info: {content[0] if content else 'Empty'}")
        print("-" * 40)
        
except Exception as e:
    print(f"Error finding group: {e}")
```
